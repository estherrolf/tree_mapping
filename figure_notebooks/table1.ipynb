```python
import json
import numpy as np
import pandas as pd

reference_maps = ['ETH', 'GLAD', 'META', 'PAULS']

with open('/n/tambe_lab/Everyone/tree_mapping/results.json') as f:
    results = json.load(f)
```

```python
metrics = ['mae', 'r2', 'rmse', 'mse']
```

```python
results_print_ref = pd.DataFrame()

for ref_map in reference_maps:
    row = pd.DataFrame(results[ref_map]['total'], index=[ref_map])
    results_print_ref = pd.concat((results_print_ref,row))
```

```python
results_print_ref[['mae', 'r2', 'rmse']].round(2)
```

```python
models = list(results.keys())
for ref_map in reference_maps: models.remove(ref_map)
```

```python
results_print_models_by_seed = pd.DataFrame()
num_trials = 10
for model in models:
    for trial in np.arange(num_trials):
        row = results[model]['12 train sites'][f'seed {trial}']['total']
        row.update({'model': model, 'trial_seed': trial})
        results_print_models_by_seed = pd.concat([results_print_models_by_seed, pd.DataFrame([row])])
        
```

```python
df_mean = results_print_models_by_seed.groupby('model').mean()
df_stddev = results_print_models_by_seed.groupby('model').std()
df_stddev = df_stddev.rename(columns= {x: f'{x} (stddev)' for x in df_stddev.columns})
df_print = df_mean.join(df_stddev)


columns_print = ['rmse', f'rmse (stddev)', 'mae', f'mae (stddev)', 'r2', f'r2 (stddev)']
models_print = ['finetune_xceptionS2/pretrained/3_layers_tuned',
                'finetune_xceptionS2/randominit_nolatlon/3_layers_tuned',
                'finetune_xceptionS2/randominit_latlon/3_layers_tuned',
                'unet/pretrained/freeze_backbone_True',
                'unet/pretrained/freeze_backbone_False',
                'unet/randominit/12_channels',
                'local_only_models/128_filters/12_channels',
                'random_forest/12_channels']

df_print.loc[models_print,columns_print].round(2)
```

```python
# appendix table for different number of xception layers tuned

base_models = ['finetune_xceptionS2/pretrained',
               'finetune_xceptionS2/randominit_nolatlon',
               'finetune_xceptionS2/randominit_latlon'
              ]
models_print_app = [[f'{base_model}/{i}_layers_tuned' for i in range(1,4)] for base_model in base_models]
models_print_app = list(np.ravel(models_print_app))
```
