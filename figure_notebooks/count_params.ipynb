```python
import torch
import numpy as np
import pandas as pd
import os
os.chdir('/n/home10/erolf/tree_mapping')
from experiment_utils import get_site_splits, read_config_file
from train_models import get_train_sites, setup_chm_datamodule
from trainers.regression_with_nans import PixelwiseRegressionTask
```

```python
cfg = read_config_file('experiment_configs/train_pretrained_xception_model.yaml')
```

```python
params_by_model_type = pd.DataFrame()
```

```python
model_type = 'FCN 12 channel'
cfg = read_config_file('experiment_configs/train_baseline_local_models.yaml')

# get this data split
split_seed = cfg['data']['split_seed']
splits = get_site_splits(split_seed)
splits_to_do = [0]
base_name = cfg['base_name']
exp_name = cfg['exp_name']
version_id_base = cfg['version_id_base']
channels = 12
cfg['num_filters'] = 128

for split_number in splits_to_do:
        task = cfg['task']
        sites_per_split = splits[split_number]
        
        # override training hyperparameters in config file if specified directly
        if lr is not None:
            task['lr'] = lr
        if weight_decay is not None:
            task['weight_decay'] = weight_decay
        if channels is not None:
            task['in_channels'] = channels
            cfg['data']['num_image_channels'] = channels
        if layers_tuned is not None:
            task['num_layers_to_unfreeze'] = layers_tuned
        if freeze_backbone is not None:
            task['freeze_backbone'] = freeze_backbone
        if train_sample_seed is not None:
            sites_per_split['train_sites'] = get_train_sites(split_seed=split_seed, split=split_number, train_sample_seed=train_sample_seed, num_train_sites=12)
        
        chm = setup_chm_datamodule(sites_per_split, cfg['data'])
        chm_task = PixelwiseRegressionTask(**task)

model = chm_task.model
pytorch_total_params = sum(p.numel() for p in model.parameters())
pytorch_trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

params_by_model_type = pd.concat((params_by_model_type, 
                                 pd.DataFrame([{'model': model_type, 
                                               'total_params': pytorch_total_params,
                                               'trainable_params': pytorch_trainable_params}])))
```
