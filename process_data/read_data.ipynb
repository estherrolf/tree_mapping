```python
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = gpd.read_file('AllProcessedSites_2022-11-03/AllProcessedSites_Server_GDrive_Updated20221103_ManualEdits_V2.shp')
```

```python
df
```

```python
np.unique(df['landscape'])
```

```python
landscapes = np.unique(df['landscape'])
fig, ax = plt.subplots(1, len(landscapes), figsize=(20,4))

for l,ls in enumerate(landscapes):    
    
    df[df['landscape'] == ls].plot(ax=ax[l])
    ax[l].set_title(ls)
    ax[l].set_aspect('equal')
    
    
```
