```python
import rasterio
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
sentinel_root_dir = '../data/raw/sentinel/'
sentinel_dirs =[os.path.join(sentinel_root_dir, x) for x in os.listdir(sentinel_root_dir)]
sentinel_dirs
```

```python
channels = ['B02', 'B03', 'B04', 'B08']
image_stats = {}
for channel in channels:
    image_stats[channel] = []
image_stats['pixel_count'] = []

for sentinel_dir in sentinel_dirs:
    for fn in os.listdir(sentinel_dir):
        if not fn.endswith('.tif'): continue
        channel = fn.split('_')[2].split('.')[0]
        if channel.startswith('B'):
            with rasterio.open(os.path.join(sentinel_dir, fn)) as f:

                data = f.read()

                this_dict = {
                    'mean': np.mean(data),
                    'std': np.std(data),
                    'max': np.max(data), 
                    'num_pix': f.shape[0] * f.shape[1],
                    'vals': data.ravel()
                }
                
                image_stats[channel].append(this_dict)
                # fig, ax = plt.subplots()
                # plt.hist(data.ravel())
                # plt.title(f'pixel values for band: {channel}')
```

```python
for channel in channels:
    print(f"{channel}: {np.mean([x['mean'] for x in image_stats[channel]])}")
```

```python
for channel in channels:
    all_pix = []
    for x in image_stats[channel]:
        all_pix.append(x['vals'])
    print(f"{channel}: {np.mean(all_pix), np.std(all_pix)}")
```
